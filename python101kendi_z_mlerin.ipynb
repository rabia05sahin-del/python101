#%S ve %D kullanımı

```python
## %S = iki metni birleştirir.
## %d = sayısal,ondalık ifadeleri birleştirmek için kullanılır.

adsoyad = "Orçun Madran"
dogumTarihi = 1976

print("Merhaba, ben %s!" % adsoyad)
print("Ben %d doğumluyum" % dogumTarihi)

ad = "Orçun"
soyad = "Madran"

print("Merhaba, ben %s %s!" % (ad, soyad))  ## burda 2 değişken olduğu için paranteze aldı.
```

While Döngüsü

**Görev: Tahmin Oyunu**

"while" döngüsü kullanarak bir tahmin oyunu tasarla. Bu tahmin oyununda, önceden belirlenmiş olan kelime ile klavyeden girilen kelime karşılaştırılmalı, tahmin doğru ise oyun "Bildiniz..!" mesajı ile sonlanmalı, yanlış ise tahmin hakkı bir daha verilmeli.

```python
#Tahmin Oyunu
kelime = "bilgi"
tahmin = ""
dogrukelime="Rabia"

input("lütfen tahmininizi giriniz.")

while dogrukelime!=tahmin:
  print("yanlış tahmin!")
  tahmin=input("lütfen tahmininizi giriniz.")

print("doğru bildiniz")
```
